```python
h = int(input("Enter number of hours: "))

if h <= 2:
    cost = h * 50

elif h <= 5:
    cost = (2 * 50) + (h - 2) * 30

else:
    cost = (2 * 50) + (3 * 30) + (h - 5) * 20

print("Total cost =", cost)
```

```python
marks = []

n = int(input("enter subject numbers: "))

for i in range(n):
    s = int(input("Enter marks:"))
    marks.append(s)

max = marks[0]
min = marks[0]
count = 0

for i in range(n):
    if marks[i] > max :
        max = marks[i]
    elif marks[i] < min:
        min = marks[i]

for j in range(n):
    if marks[j] > 80:
        count = count + 1


print("Marks:",marks)
print("First 3 subjects:",marks[:3])
print("Last 2 subjects:",marks[-2:])
print("highest mark:",max)
print("Lowest mark:",min)
print("Count greater than 80:",count)

```

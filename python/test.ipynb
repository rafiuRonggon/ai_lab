```python
a = int(input("enter number: "))
b = int(input("enter number: "))
operator = input("enter operator: ")

if(operator == "+"):
    print(a + b)
elif(operator == "-"):
    print(a - b)
elif(operator == "*"):
    print(a * b)
elif(operator == "/"):
    print(a / b)
else:
    print("not a valid operator")
```

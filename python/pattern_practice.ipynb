```python
N = int(input("Sample input: "))

# Outer loop runs N times, once for each row
for i in range(1, N + 1):
      # Inner loop prints 'N - i' spaces
    for j in range(1, N - i + 1):
        print(" ", end="")
        
    # Inner loop prints '2 * i - 1' stars
    for j in range(1, 2 * i):
        print("*", end="")
    # Move to the next line
    print()
```

```python
def printRows(row, n, upper):
    if row < 1 or row > n:
        return

    # Print leading spaces.
    for j in range(n - row):
        print(" ", end="")

    # Print stars.
    for j in range(1, row + 1):
        print("* ", end="")

    print()

    if upper:
        printRows(row + 1, n, upper)

        if row == n:
            printRows(n, n, False)
    else:
        printRows(row - 1, n, upper)

def printDiamond(n):
    printRows(1, n, True)

n = int(input("Sample input: "))

printDiamond(n)
```

```python
n = int(input("sample input: "))

for i in range(2 * n - 1):
    if i < n:
        comp = 2 * (n - i) - 1
    else:
        comp = 2 * (i - n + 1) + 1

    # Print leading whitespaces
    print(" " * (2 * n - comp - 1), end="")

    # Print stars *
    print("* " * comp)
```

```python
n = int(input("sample input: "))

spaces = 2 * n - 1
stars = 0

for i in range(1, 2 * n):

    if i <= n:
        stars += 1
        spaces -= 2
    else:
        stars -= 1
        spaces += 2

    # Left side stars
    for j in range(1, stars + 1):
        print("*", end="")

    # Middle spaces
    for j in range(1, spaces + 1):
        print(" ", end="")

    # Right side stars
    for j in range(1, stars + 1):
        if j != n:
            print("*", end="")

    print()
```

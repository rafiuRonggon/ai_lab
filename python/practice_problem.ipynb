```python
x = int(input("sample input: "))
sum_even = 0
even_count = 0
sum_odd = 0
odd_count = 0

for x in range(1, 11):
    if x % 2 == 0:
        sum_even = sum_even + x
        even_count = even_count + 1
    elif x % 2 != 0:
        sum_odd = sum_odd + x
        odd_count = odd_count + 1

print("Sample Output: ")
print("even sum: ", sum_even)
print("odd sum: ", sum_odd)
print("even count: ", even_count)
print("odd count: ", odd_count)
```

```python
x = int(input("sample input: "))
last = x % 10
total_digit = 0
sumOfDigit = 0
temp = x
while temp > 0:
    current_digit = temp % 10
    sumOfDigit = current_digit + sumOfDigit
    total_digit = total_digit + 1
    temp = temp // 10

print("Sample Output: ")
print("Total digits: ", total_digit)
print("Sum of digits: ", sumOfDigit)
print("Last digit: ", last)

```

```python
marks = []
print("Sample Input:")
total = 0

for i in range(1, 7):
    mark = int(input(f"{i}: "))
    marks.append(mark)

for value in marks:
    total += value
avg = total / len(marks)
pass_count = 0
for value in marks:
    if value >= 50:
        pass_count += 1

print("Sample Output:")
print("First 3:", marks[:3])
print("Last 3:", marks[3:6])
print("Total:", total)
print("Pass Count:", pass_count)
print("Average:", f"{avg:.2f}")
```

```python
numbers = []
for i in range(1,9):
    numbers.append(int(input(f"{i}: ")))

positive = 0
negative = 0
zero = 0
for value in numbers:
    if value > 0:
        positive += 1
    elif value < 0:
        negative += 1
    else:
        zero += 1

print("Sample Output:")
print("numbers:", numbers)
print("Positive:", positive)
print("Negative:", negative)
print("Zero:", zero)
```

```python
numbers = []
for i in range(5):
    numbers = int(input("enter a number: "))
    numbers.append(numbers)
max = numbers[0]
min = numbers[0]
for i in range(1, len(numbers)):
    if numbers[i] > max:
        max = numbers[i]
    elif numbers[i] < min:
        min = numbers[i]
difference = max - min
print("Sample Output:")
print("largest:",max)
print("smallest:",min)
print("difference:",difference)
```

```python
text = input("sample input: ")
vowels = 0
consonants = 0
for i in range(len(text)):
    ch = text[i].lower()
    if ch in 'aeiou':
        vowels += 1
    elif ch.isalpha():
        consonants += 1

print("Sample Output:")
print("vowels:", vowels)
print("consonants:", consonants)
print("reversed:", text[::-1])
```

```python
n =int(input("enter numbers of products: "))
price  = []

for i in range(n):
    p = int(input("enter price: "))
    price.append(p)

total = 0
count = 0

for i in range(n):
    total = total + price[i]

    if price[i] > 500:
        count +=1
if total < 1000:
    discount = 0
elif total < 3000:
    discount = total*0.05
else:
    discount = total*0.1

lastPrice = total - discount

print("sample output: ")
print("Total bill: ",total)
print("Product Above 500: ",count)
print("discount: ",int(discount))
print("Final Bill: ",int(lastPrice))

    
```
